# Setup & Quickstart — get a green cell in under 2 minutes

This notebook gets you from zero to your first OpenAI **Responses API** call as fast as possible. Install, load your key, fire one request, see output.

```python
import os
from dotenv import load_dotenv
from openai import OpenAI

# Loads OPENAI_API_KEY from a .env file in the project root.
load_dotenv()

client = OpenAI()  # reads OPENAI_API_KEY from the environment
```

```python
response = client.responses.create(
    model="gpt-5.4-mini",
    input="Say hello and tell me one fun fact about the Python language.",
)
print(response.output_text)
```

## What just happened?

- `client.responses.create(...)` is the **Responses API** — the current OpenAI interface that replaces `chat.completions.create`.
- `input=` takes a plain string (a single user message) or a list of message dicts.
- `response.output_text` is the SDK helper that flattens the model's text output.

```python
# Multi-turn: chain a follow-up with previous_response_id (no manual history re-send).
follow_up = client.responses.create(
    model="gpt-5.4-mini",
    previous_response_id=response.id,
    input=[{"role": "user", "content": "Now say it again, but as a haiku."}],
)
print(follow_up.output_text)
```

# PDF / Invoice Extraction with Pydantic — Responses API

We extract structured invoice fields using the native `client.responses.parse(text_format=MyModel)` path, which returns a parsed Pydantic instance on `response.output_parsed`. One optional cell shows the `instructor` library for multi-provider portability.

```python
import os
from dotenv import load_dotenv
from openai import OpenAI

# loading from a .env file
# load_dotenv(dotenv_path="/full/path/to/your/.env")

# or 
# if you're on google colab just uncomment below and replace with your openai api key
# os.environ["OPENAI_API_KEY"] = "<your-openai-api-key>"
```

```python
from pydantic import BaseModel, Field
from openai import OpenAI

client = OpenAI()

class InvoiceData(BaseModel):
    """Structured invoice fields extracted from a PDF/text."""
    invoice_number: str = Field(..., description="The invoice number extracted from the PDF")
    vat_registration_number: str = Field(..., description="The VAT registration number extracted from the PDF")

fake_invoice_data = """
Invoice Summary
==============
Invoice No: INV-123456
Date: 2024-02-11
Billing To: ACME Corporation
Address: 123 Business Rd, Business City, BC1234

Product Description       Quantity    Unit Price    Total
-----------------------------------------------------------
Widget A                  10          $15.00        $150.00
Gadget B                  5           $20.00        $100.00

Subtotal:                                          $250.00
Tax (10%):                                          $25.00
Total:                                             $275.00

VAT Registration Number: VAT-7890123

Thank you for your business!
"""

# Native Responses API structured output via responses.parse(text_format=...).
response = client.responses.parse(
    model="gpt-5.4-mini",
    instructions="You extract structured invoice data from raw invoice text.",
    input=f"Invoice: {fake_invoice_data}\nExtract the invoice data.",
    text_format=InvoiceData,
)
output_invoice = response.output_parsed
```

```python
# Optional: multi-provider portability with instructor.
import instructor

instructor_client = instructor.from_provider(
    "openai/gpt-5.5",
    mode=instructor.Mode.RESPONSES_TOOLS,
)

invoice_via_instructor = instructor_client.responses.create(
    input=f"Invoice: {fake_invoice_data}. Extract the invoice data.",
    response_model=InvoiceData,
)
print(invoice_via_instructor)
```

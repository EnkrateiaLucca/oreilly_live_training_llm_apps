## Structured tool/function definitions with Pydantic — Responses API

Instead of LangChain's deprecated `convert_pydantic_to_openai_function`, we use the native Responses API: define a Pydantic model and either (a) extract structured data with `responses.parse(text_format=...)`, or (b) expose it as a function tool via `responses.create(tools=[...])`.

```python
from pydantic import BaseModel, Field
from openai import OpenAI

client = OpenAI()


class WeatherSearch(BaseModel):
    """The weather query for a specific airport."""
    airport_code: str = Field(description="The airport code of the airport you want the weather of.")
```

SO the airport_code is something that the LLM will figure out on how to input to the function, therefore making the call with the correct input.

```python
response = client.responses.parse(
    model="gpt-5.4-mini",
    instructions="You map a city weather request to its airport code.",
    input="What's the weather like in Lisbon today?",
    text_format=WeatherSearch,
)
response.output_parsed
```

We can see from below that the LLM figured out that the airport_code for Lisbon is `LIS`.

You can also bind the function to the model to avoid having to pass the `functions` keyword argument everytime. 

```python
# As a function tool the model can call (flat Responses API tool shape).
weather_tool = {
    "type": "function",
    "name": "weather_search",
    "description": "Get the weather of a specific airport.",
    "parameters": WeatherSearch.model_json_schema(),
}

response = client.responses.create(
    model="gpt-5.4-mini",
    input="What's the weather like in Lisbon today?",
    tools=[weather_tool],
)
response.output
```

YOu can also force the model to use the function by binding it with it and adding the `function_call` keyword argument to a dictionary containing the key `name` and the name of that function `WeatherSearch`:

```python
# Force the model to call the tool with tool_choice.
response = client.responses.create(
    model="gpt-5.4-mini",
    input="What's the weather like in Lisbon today?",
    tools=[weather_tool],
    tool_choice={"type": "function", "name": "weather_search"},
)
response.output
```

Now, let's connect to the stuff we already know.

```python
response = client.responses.create(
    model="gpt-5.4-mini",
    instructions="You are a helpful assistant that tells the weather.",
    input="What's the weather like in Lisbon today?",
    tools=[weather_tool],
    tool_choice={"type": "function", "name": "weather_search"},
)
response.output
```

For multiple functions:

```python
from typing import List

class SearchExamples(BaseModel):
    """Search for examples of a concept."""
    concept: str = Field(description="The concept you want to search for.")

class AnalogySearch(BaseModel):
    """Search for analogies of a concept."""
    concept: str = Field(description="The concept you want to search for.")


example_tool = {
    "type": "function",
    "name": "search_examples",
    "description": "Search for examples of a concept.",
    "parameters": SearchExamples.model_json_schema(),
}
analogy_tool = {
    "type": "function",
    "name": "analogy_search",
    "description": "Search for analogies of a concept.",
    "parameters": AnalogySearch.model_json_schema(),
}
```

```python
tools = [analogy_tool, example_tool]
```

```python
# Pass `tools` directly on each responses.create call below.
```

```python
response = client.responses.create(
    model="gpt-5.4-mini",
    input="List examples for what an SDK is.",
    tools=tools,
)
response.output
```

```python
response = client.responses.create(
    model="gpt-5.4-mini",
    input="List analogies for what an SDK is.",
    tools=tools,
)
response.output
```

## Optional: multi-provider portability with `instructor`

`instructor` works with the Responses API and gives you the same Pydantic ergonomics across providers.

```python
import instructor

instructor_client = instructor.from_provider(
    "openai/gpt-5.5",
    mode=instructor.Mode.RESPONSES_TOOLS,
)

weather = instructor_client.responses.create(
    input="What's the weather like in Lisbon today?",
    response_model=WeatherSearch,
)
print(weather)
```
